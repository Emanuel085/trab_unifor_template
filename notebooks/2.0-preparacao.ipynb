# CRISP-DM - Etapa 03: Preparação dos Dados
## Tema 10: Indústria, Energia e Meio Ambiente (AI4I 2020 Predictive Maintenance)

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder, StandardScaler

# carregar base
DATA_PATH = "../data/raw/ai4i2020.csv"
df = pd.read_csv(DATA_PATH)

# isolar identificadores e rotulos reais de falha (regra PBL)
colunas_remover = ["UDI", "Product ID", "Machine failure", "TWF", "HDF", "PWF", "OSF", "RNF"]
colunas_existentes = [c for c in colunas_remover if c in df.columns]

df_features = df.drop(columns=colunas_existentes).copy()
df_labels = df[colunas_existentes].copy()

# codificacao ordinal para variavel 'Type' (L, M, H)
encoder = OrdinalEncoder(categories=[["L", "M", "H"]])
df_features["Type_encoded"] = encoder.fit_transform(df_features[["Type"]])
df_features = df_features.drop(columns=["Type"])

# escalonamento das variaveis numericas
scaler = StandardScaler()
colunas_num = df_features.columns

df_scaled = pd.DataFrame(scaler.fit_transform(df_features), columns=colunas_num)

print("Estrutura dos dados preparados e escalonados:")
df_scaled.head()
```

```python
# guardar o dataset processado para a fase de modelagem
df_scaled.to_parquet("../data/processed/ai4i2020_processed.parquet", index=False)
print("Dados salvos em data/processed/ai412020_processed.parquet")
```
